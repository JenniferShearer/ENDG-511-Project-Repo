```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ollioctopus/asl-dataset")

print("Path to dataset files:", path)
```

```python
#display images

import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# The 'path' variable points to a directory, not an image file.
# We need to find an actual image file within this directory or its subdirectories.

image_file_path = None
# Walk through the directory to find the first image file
for root, dirs, files in os.walk(path):
    for file in files:
        if file.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp')):
            image_file_path = os.path.join(root, file)
            break
    if image_file_path:
        break

if image_file_path:
    img = mpimg.imread(image_file_path)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
else:
    print(f"No image files found in the directory: {path}")
```
